# 3. Functions

```python
def zero_shot_prompt(text):

    intro = (
        "Classify the sentiment of the following movie review from the IMDB dataset.\n"
        "Respond only with a single digit: 0 if the sentiment is negative, "
        "1 if the sentiment is positive.\n"
    )

    target = f"Review: \"{text}\"\nLabel:"

    return intro + target
```

```python
def predict_label(text, prompt):

    try:

        start_time = time.perf_counter()
        first_token_time = None
        output_text = ""
        usage = None

        with client.messages.stream(
            model="claude-sonnet-4-5-20250929",
            max_tokens=16,
            temperature=0.0,
            messages=[{"role": "user", "content": prompt}],
        ) as stream:
            for event in stream:
                if event.type == "content_block_delta":
                    delta_text = getattr(event.delta, "text", None)
                    if delta_text:
                        if first_token_time is None:
                            first_token_time = time.perf_counter()
                        output_text += delta_text

            final_msg = stream.get_final_message()
            usage = getattr(final_msg, "usage", None)

        end_time = time.perf_counter()

        return {
            "prediction": output_text.strip(),
            "latency_ms": (end_time - start_time) * 1000,
            "ttft_ms": ((first_token_time - start_time) * 1000) if first_token_time else None,
            "input_tokens": getattr(usage, "input_tokens", None) if usage else None,
            "output_tokens": getattr(usage, "output_tokens", None) if usage else None,
        }

    except:

        return {
            "prediction": '-1',
            "latency_ms": '-',
            "ttft_ms": '-',
            "input_tokens": '-',
            "output_tokens": '-',
        }
```

# 4. Load Dataset

```python
df = pd.read_csv(path_open)
```

```python
df.shape
```

```python
pred_label = []
pred_latency = []
pred_ttft = []
pred_input_tokens = []
pred_output_tokens = []
```

```python
for i in range(len(df)):

  text = df['text'].iloc[i]
  prompt = zero_shot_prompt(text)
  output = predict_label(text, prompt)

  try:
    final_output = int(output['prediction'])
    final_latency = output['latency_ms']
    final_ttft = output['ttft_ms']
    final_input_tokens = output['input_tokens']
    final_output_tokens = output['output_tokens']
  except:
    final_output = '-1'
    final_latency = '-'
    final_ttft = '-'
    final_input_tokens = '-'
    final_output_tokens = '-'

  pred_label.append(final_output)
  pred_latency.append(final_latency)
  pred_ttft.append(final_ttft)
  pred_input_tokens.append(final_input_tokens)
  pred_output_tokens.append(final_output_tokens)

  time.sleep(1)

  if (i % 10) == 0:
    print(i)
```

```python
df[f'{name_model}-{mode}-seed-{seed}-label'] = pred_label
df[f'{name_model}-{mode}-seed-{seed}-latency'] = pred_latency
df[f'{name_model}-{mode}-seed-{seed}-ttft'] = pred_ttft
df[f'{name_model}-{mode}-seed-{seed}-input-tokens'] = pred_input_tokens
df[f'{name_model}-{mode}-seed-{seed}-output-tokens'] = pred_output_tokens
```
